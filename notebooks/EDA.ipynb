# 📊 Exploratory Data Analysis — Water Quality Dataset

This notebook explores the synthetic water quality dataset used to train
our LSTM model. We look at distributions, correlations, time-series
trends, and the Safe/Unsafe class balance.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='darkgrid')
%matplotlib inline
```

## 1. Load Dataset

```
df = pd.read_csv('../data/water_quality.csv')
print(f'Shape: {df.shape}')
df.head()
```

```
df.info()
```

```
df.describe()
```

## 2. Missing Values

```
missing = df.isnull().sum()
print(missing[missing > 0])
print(f'\nTotal missing: {df.isnull().sum().sum()}')
```

## 3. Feature Distributions

```
features = ['pH', 'Turbidity', 'Dissolved_Oxygen', 'Temperature', 'Conductivity', 'WQI']
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.flatten(), features):
    df[col].hist(bins=50, ax=ax, color='steelblue', edgecolor='white')
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

## 4. Correlation Heatmap

```
corr = df[features].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('Feature Correlation Matrix')
plt.show()
```

## 5. Time-Series Trends

```
df['timestamp'] = pd.to_datetime(df['timestamp'])
fig, axes = plt.subplots(3, 2, figsize=(15, 10))
for ax, col in zip(axes.flatten(), features):
    ax.plot(df['timestamp'][:500], df[col][:500], linewidth=0.8)
    ax.set_title(col)
    ax.set_xlabel('Time')
plt.tight_layout()
plt.show()
```

## 6. Safe vs Unsafe Distribution

```
counts = df['Status'].value_counts()
plt.figure(figsize=(6, 4))
counts.plot(kind='bar', color=['#4CAF50', '#F44336'])
plt.title('Water Quality Status Distribution')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
print(counts)
```
